```python
def first_scroll_data(data, name_dataset):
    '''Функция для просмотра базовой информации о
    датасете'''
    line_length = 150
    print('-' * line_length)
    print(f'Обзор датасета: {name_dataset}'.center(line_length))
    print('-' * line_length)


    print('Общая информация:')
    print(data.info())
    print()
    print('Первые пять строк таблицы:')
    display(data.head())
    print()
    print('Последние пять строк таблицы:')
    display(data.tail())

    print()
    print()
```

```python
def list_scroll(data: list, name_dataset:list):
    '''вторичная функция для прокрутки first_scroll_data 
    используя список'''
    cnt = len(name_dataset)
    number = 0
    for i in data:
        first_scroll_data(i, name_dataset[number])
        number += 1
        print()
```
