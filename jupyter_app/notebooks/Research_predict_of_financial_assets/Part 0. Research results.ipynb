## Блок исследования применения машинного обучения в пердсказании стоимости финансовых активов


Данная работа является исследовательской частью магистерской работы "Разработка универсального сервиса для прогнозирования финансовых активов с применением методов машинного обучения" и сфокусирована на анализе и прогнозировании временных рядов различных финансовых активов

### Цель исследования

**Цель исследования** — оценить в каких сферах лучше всего модели предсказания справляются с поставленной задачей, построить различные варианты моделей (машинного обучения и классических статистических моделей), способные предсказывать будущие значения цен на основе их исторических данных.

### Структура работы

Работа строится по следующей схеме:
- Каждый блок работы вынесен в отдельную тетрадку для удобства просмотра данных и удобства корректировки работы в целом
- Создана 1 тетрадка (Part 0. Research results.ipynb) в которой будут сделаны основные выводы по проведенным исследованиям в каждом блоке
- В каждом блоке исследований будет оглавление, с краткой информацией о проводимых в данном блоке исследований и минимальным кол-вом текста.

## *Итоги исследования [Part 1](./Part%201.%20Load%20data.ipynb). Загрузка и предобработка данных*,

Основное исслендование будет проводиться на 10 датасетах с временными рядами различных финансовых активов, а именно:

- Криптовалюта
- Нефтеные фьючерсы
- Акции метталургических компаний
- Акции нефтедобывающих компаний
- Спотовы цена на драгоценные металы


### Общая характеристика для всех датасетов
Обнаружено небольшое кол-во пропусков в данных а также текстовый тип всех колонок. Местами присутствуют ошибочные строки, появившиеся при сохранении запаршенных данных из библиотеки yfinance. Для исследования были выбраны данные за Даннные за 2016-2024 (2025) год. В 2025 год в основном будет выдаваться как тестовая часть для тестирования результатов предсказания моделей. Временной промежуток в 8–9 лет позволяет захватить как краткосрочные, так и долгосрочные тренды, сезонность, повторяющиеся паттерны и случайные колебания. Так же в выбранный интервал входят важные экономические события, включая:
- Посткризисный период после 2014 года;
- Пандемия COVID-19 (2020–2021);
- Геополитическая и экономическая турбулентность последних лет;
- Рост волатильности на финансовых рынках (2022–2023).


#### Характеристика датасетов "Криптовалюта"

Было рассмотрено 2 датасета со стоимостью Ethereum и BTC. Данные активы были выбраны т.к. они являются наиболее популярноми в сфере криптовалюты.
В ходе загрузки данных в датасете Ethereum - 3293 строчек, в BTC - 3361.


#### Характеристика датасетов "Нефтеные фьючерсы" 

Фьючерсы на нефть марки брент с выкупом через 5 месяцев Brent и WTI были выбраны т.к. они являются крупным энергетическим активом, пользующимся популярностью в Европе и Америке. В обоих датасетах ~2376 строчки. Такое кол-во данных объясняется тем что фьчерсы торгуются с времеными перерывами.


#### Характеристика датасетов "Акции метталургических компаний" 

Для исследования были выбраны акции 2вух крупных Китайский предприятия jiangxi copper - специализируется на продаже цветных металлов и  baoshan_iron - крупнейшее сталелитейное предприятие. Тут в каждом датасете ~2283 строчки. Так же видно что акциями торгавали с перерывами по времению


#### Характеристика датасетов "Акции нефтедобывающих компаний" 

Для исследования были выбраны акции petrochina и equinor oslo (Китайска и Норвежская компании) торгующие нефтью. Тут аналогичная ситуация как и с метталургическими компании. Всего в каждом датасете ~ 2332 строчки, но etrochina за всё время торговалась на 50 дней меньше.


#### Характеристика датасетов "Спотовы цена на драгоценные металы" 

Один из самых устоявшихся активов для хранения накоплений в натуральном виде - драгоценные металлы. Для исследования были выбраны спотовые цены на золото и серебро. Всего в каждом датасете ~2395 строчки.


**Итог:** Были загружены и просмотерны все датасеты. В большинстве своём количество данных одинаково в каждой группе по отдельности. Есть некорректные типы данных и значения в датасетах. Присутствие пустых данных минимально. Сформирована функция datasets_load.py для дальнейшей загрузки данных в разных тетрадках.

### Итоги исследования датасетов "Криптовалюта"

##### Общие итоги для датасетов
- На графике временного ряда наблюдается чётко выраженный восходящий тренд Ковидного периода, псоле конца которого стоимость резко сократилась ввиду снижения популярности майнинга, так же виден существенные колебания цены по времени.
- По корреляции новых фичей видно, что разность цен в различных временных лагах по всем ценам слабо коррелирует со всеми фичами. График автокорреляции показывает резкое снижение с увеличением числа лагов. Наблюдается высокая положительная автокорреляция при малых лагах, что указывает на сильную зависимость текущих значений от небольшого кол-ва прошлых. График ACF показывает, что значения временного ряда имеют устойчивую положительную автокорреляцию до 50 лага.
- В периоды бурного роста или падения (2021–2022) волатильность резко возрастает. Ярче всего волатильность просматривается на лаге в 15 дней. Менее яркая волатильность в сдвиге на 5 дней. Исключение - волатильность минимальной цены, там 5 дневный и 15 дневный сдвиг практически идентичны друг другу, а значит цена октрытия имеет высокую волатильность.
- Ряд не стационарный
- **В сравнении с эжфиром, биткоин имеет более стабильную структуру графика в виду большего его признания на мировом уровне как универсальной монеты.**

##### Выводы по исследованию датасета crypt_Ethereum:

### Итоги исследования датасетов "Нефтеные фьючерсы" 

##### Общие итоги для датасетов
- Ряды выглядят более стабильным по сравнению с криптовалютами, но в них присутствуют сильные падения цены связанные с изменением в энергетической политике различных государств закупающих нефть.
- Тренд — ярко выраженный рост в 2021–2022 году, далее постепенное снижение и стабилизация. Сезонность — выражена слабо, но периодическая структура сохраняется. Особенно сильные шумовые колебания в начале 2022 года.

Фичи в виде разности цен в различных временных лагах по всем ценам слабо коррелирует со всеми фичами. Так же видно что стандартное отклонение аналогично слабо корреллирует со всеми фичами.
- Волатильность резко увеличивается в пиковые периоды (2022 год) затем снижается. Ярче всего волатильность просматривается на лагах в 15 дней. Менее яркая волатильность в сдвиге на 5 дней.

- Z-score выявил всплески в 2022 году, когда цена превысила порог в 3 стандартных отклонения. IQR подтвердил те же участки как выбросы — особенно в марте–июне 2022 года.
